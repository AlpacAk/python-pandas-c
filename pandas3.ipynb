```python
import pandas as pd
import os

import kagglehub


path = kagglehub.dataset_download("udaymalviya/bank-loan-data")

print("Path to dataset files:", path)
DATAPATH = path
df = pd.read_csv(os.path.join(DATAPATH, 'loan_data.csv'))

df.rename(columns={'loan_amnt': 'loan_amount'}, inplace=True)
# tylko kolumny 'person_gender', 'person_age' oraz 'person_income'
# wiersze od 1000 do 2000 włącznie
# wiersze parzyste
# wiersze o parzystej wartości w kolumnie indeks
# pierwszą i ostatnią kolumnę
a = df[['person_gender', 'person_age', 'person_income']]
b = df.iloc[999:2000]
c = df.iloc[::2]
d = df[df.index % 2 == 0]
e = df.iloc[:, [0, -1]]
display(a, b, c, d, e)
# tylko kobiety z wykształceniem 'Master' oraz 'Bachelor' i pożyczkami pomiędzy 50000 a 70000
# wszyscy mężczyźni z wykształceniem innym niż 'Master' oraz będący właścicielem lokalu
# wszystkie osoby z wartością loan_percent_income w przedziale od 0.4 do 0.6
# wszystkie wartości pożyczek, których kwota jest wielokrotnością liczby 10000
df[(df['person_gender'] == 'female') & (df['person_education'].isin(['Master', 'Bachelor'])) & (df['loan_amount'] >= 50000) & (df['loan_amount'] <= 70000)]
df[(df['person_gender'] == 'male') & (df['person_education'] != 'Master') & (df['person_home_ownership'] == 'OWN')]
df[(df['loan_percent_income'] >= 0.4) & (df['loan_percent_income'] <= 0.6)]
df[df['loan_amount'] % 10000 == 0]
```
