```python
import time

MOVES = [
    (2, 1), (1, 2), (-1, 2), (-2, 1),
    (-2, -1), (-1, -2), (1, -2), (2, -1)
]

def is_valid(x, y, board, n):
    return 0 <= x < n and 0 <= y < n and board[x][y] == -1

class KnightsTourBasic:

    def __init__(self, n):
        self.n = n
        self.steps = 0

    def solve(self, start_x, start_y):
        board = [[-1 for _ in range(self.n)] for _ in range(self.n)]
        board[start_x][start_y] = 0
        if self._backtrack(start_x, start_y, 1, board):
            return board
        return None

    def _backtrack(self, x, y, move_i, board):
        self.steps += 1
        if move_i == self.n * self.n:
            return True

        for dx, dy in MOVES:
            nx, ny = x + dx, y + dy
            if is_valid(nx, ny, board, self.n):
                board[nx][ny] = move_i
                if self._backtrack(nx, ny, move_i + 1, board):
                    return True
                board[nx][ny] = -1
        return False

def count_onward_moves(x, y, board, n):
    count = 0
    for dx, dy in MOVES:
        nx, ny = x + dx, y + dy
        if is_valid(nx, ny, board, n):
            count += 1
    return count

class KnightsTourWarnsdorff:

    def __init__(self, n):
        self.n = n
        self.steps = 0

    def solve(self, start_x, start_y):
        board = [[-1 for _ in range(self.n)] for _ in range(self.n)]
        board[start_x][start_y] = 0
        if self._backtrack(start_x, start_y, 1, board):
            return board
        return None

    def _backtrack(self, x, y, move_i, board):
        self.steps += 1
        if move_i == self.n * self.n:
            return True

        candidates = []
        for dx, dy in MOVES:
            nx, ny = x + dx, y + dy
            if is_valid(nx, ny, board, self.n):
                c = count_onward_moves(nx, ny, board, self.n)
                candidates.append((c, nx, ny))

        candidates.sort(key=lambda x: x[0])

        for _, nx, ny in candidates:
            board[nx][ny] = move_i
            if self._backtrack(nx, ny, move_i + 1, board):
                return True
            board[nx][ny] = -1
        return False

def print_board(board):
    n = len(board)
    for i in range(n):
        for j in range(n):
            print(f"{board[i][j]:3}", end=" ")
        print()

n = 6
start_x = 0
start_y = 0

basic_solver = KnightsTourBasic(n)
start_time = time.time()
solution_basic = basic_solver.solve(start_x, start_y)
end_time = time.time()

if solution_basic:
    print("Bez heurystyki:")
    print_board(solution_basic)
else:
    print("Brak rozwiązania")

print("Kroki:", basic_solver.steps)
print("Czas:", end_time - start_time)

print()

warn_solver = KnightsTourWarnsdorff(n)
start_time = time.time()
solution_warn = warn_solver.solve(start_x, start_y)
end_time = time.time()

if solution_warn:
    print("Warnsdorff:")
    print_board(solution_warn)
else:
    print("Brak rozwiązania")

print("Kroki:", warn_solver.steps)
print("Czas:", end_time - start_time)
```
