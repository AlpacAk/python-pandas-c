```python
import math

def print_board(board):
    for i in range(3):
        print(board[i*3:(i+1)*3])
    print()

def check_winner(board):
    lines = [
        (0,1,2),(3,4,5),(6,7,8),
        (0,3,6),(1,4,7),(2,5,8),
        (0,4,8),(2,4,6)
    ]
    for a,b,c in lines:
        if board[a] != " " and board[a] == board[b] == board[c]:
            return board[a]
    if " " not in board:
        return "draw"
    return None

def get_moves(board):
    return [i for i in range(9) if board[i] == " "]

def minimax(board, depth, is_max, max_depth, counter):
    counter[0] += 1
    winner = check_winner(board)

    if winner == "O":
        return 1
    if winner == "X":
        return -1
    if winner == "draw":
        return 0
    if depth == max_depth:
        return 0

    if is_max:
        best = -math.inf
        for move in get_moves(board):
            board[move] = "O"
            val = minimax(board, depth+1, False, max_depth, counter)
            board[move] = " "
            best = max(best, val)
        return best
    else:
        best = math.inf
        for move in get_moves(board):
            board[move] = "X"
            val = minimax(board, depth+1, True, max_depth, counter)
            board[move] = " "
            best = min(best, val)
        return best

def minimax_ab(board, depth, is_max, max_depth, alpha, beta, counter):
    counter[0] += 1
    winner = check_winner(board)

    if winner == "O":
        return 1
    if winner == "X":
        return -1
    if winner == "draw":
        return 0
    if depth == max_depth:
        return 0

    if is_max:
        best = -math.inf
        for move in get_moves(board):
            board[move] = "O"
            val = minimax_ab(board, depth+1, False, max_depth, alpha, beta, counter)
            board[move] = " "
            best = max(best, val)
            alpha = max(alpha, best)
            if beta <= alpha:
                break
        return best
    else:
        best = math.inf
        for move in get_moves(board):
            board[move] = "X"
            val = minimax_ab(board, depth+1, True, max_depth, alpha, beta, counter)
            board[move] = " "
            best = min(best, val)
            beta = min(beta, best)
            if beta <= alpha:
                break
        return best

def best_move(board, algorithm, max_depth):
    best_val = -math.inf
    move_choice = None
    counter = [0]

    for move in get_moves(board):
        board[move] = "O"
        if algorithm == "minimax":
            move_val = minimax(board, 0, False, max_depth, counter)
        else:
            move_val = minimax_ab(board, 0, False, max_depth, -math.inf, math.inf, counter)
        board[move] = " "

        if move_val > best_val:
            best_val = move_val
            move_choice = move

    return move_choice, counter[0]

board = [" "] * 9

algorithm = "alphabeta"
max_depth = 9

while True:
    print_board(board)
    if check_winner(board):
        break

    move = int(input("Twój ruch (0-8): "))
    if board[move] != " ":
        continue
    board[move] = "X"

    if check_winner(board):
        break

    ai_move, steps = best_move(board, algorithm, max_depth)
    print("AI rozważyło stanów:", steps)
    board[ai_move] = "O"

winner = check_winner(board)
print_board(board)

if winner == "draw":
    print("Remis")
else:
    print("Wygrywa:", winner)
```
